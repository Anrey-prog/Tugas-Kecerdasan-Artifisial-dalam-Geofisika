# RECIPE AI: STUDI KASUS GEOFISIKA METODE GEOLISTRIK

## 1. Pendahuluan

Metode geolistrik resistivitas merupakan metode geofisika yang digunakan untuk mempelajari kondisi bawah permukaan berdasarkan perbedaan sifat hambatan listrik material. Metode ini dapat digunakan untuk mengidentifikasi lapisan tanah dan batuan, memperkirakan potensi akuifer air tanah, serta menginvestigasi zona rekahan dan kondisi geologi bawah permukaan.

Recipe ini dirancang untuk membantu proses analisis data geolistrik secara sistematis menggunakan Python.

## 2. Tujuan

1. Menghitung nilai resistivitas semu berdasarkan data arus dan beda potensial.
2. Memeriksa kualitas dan konsistensi data pengukuran.
3. Memvisualisasikan distribusi resistivitas semu.
4. Mendukung interpretasi awal kondisi bawah permukaan.
5. Menyediakan alur kerja yang dapat dikembangkan untuk pemodelan inversi geolistrik.

## 3. Studi Kasus

Studi kasus yang digunakan adalah identifikasi potensi akuifer air tanah menggunakan metode geolistrik resistivitas.

Analisis dilakukan untuk mengidentifikasi variasi resistivitas bawah permukaan dan memperkirakan zona yang berpotensi mengandung air tanah. Interpretasi harus mempertimbangkan kondisi geologi, litologi, porositas, kandungan lempung, serta salinitas air.

Apabila data lapangan belum tersedia, analisis dapat dimulai menggunakan data simulasi yang dinyatakan secara jelas sebagai data hipotetis.

## 4. Data Masukan

Data yang diperlukan meliputi:

- Arus listrik (I), dalam ampere.
- Beda potensial (ΔV), dalam volt.
- Faktor geometri elektroda (K), dalam meter.
- Posisi elektroda dan konfigurasi pengukuran.
- Informasi geologi pendukung, jika tersedia.

## 5. Persamaan Dasar

Resistivitas semu dihitung menggunakan persamaan:

ρa = K × (ΔV/I)

Keterangan:

- ρa = resistivitas semu (Ω m).
- K = faktor geometri elektroda (m).
- ΔV = beda potensial (V).
- I = arus listrik (A).

Faktor geometri harus sesuai dengan konfigurasi elektroda yang digunakan.

## 6. Tahapan Recipe

### Tahap 1: Persiapan Data
Memasukkan data pengukuran dan memeriksa satuan, nilai kosong, serta kemungkinan kesalahan pencatatan.

### Tahap 2: Pengolahan Data
Menghitung resistivitas semu berdasarkan arus, beda potensial, dan faktor geometri.

### Tahap 3: Visualisasi
Membuat grafik distribusi resistivitas semu berdasarkan posisi atau nomor pengukuran.

### Tahap 4: Interpretasi
Mengidentifikasi variasi resistivitas dan membahas kemungkinan hubungannya dengan kondisi geologi bawah permukaan.

### Tahap 5: Validasi
Mengevaluasi hasil dengan data geologi, informasi sumur, atau hasil inversi apabila tersedia.

### Tahap 6: Pelaporan
Menyusun hasil pengolahan, grafik, interpretasi, keterbatasan, dan rekomendasi penelitian lanjutan.

## 7. Output

Program menghasilkan:

1. Tabel hasil perhitungan resistivitas semu.
2. Grafik resistivitas semu.
3. Ringkasan statistik data.
4. Berkas CSV hasil pengolahan.

## 8. Keterbatasan

Resistivitas semu bukan nilai resistivitas sebenarnya dari satu lapisan tertentu. Penentuan struktur bawah permukaan memerlukan pemodelan inversi yang sesuai dengan konfigurasi dan geometri pengukuran. Identifikasi akuifer juga tidak dapat dipastikan hanya berdasarkan nilai resistivitas, sehingga diperlukan informasi geologi pendukung dan validasi lapangan.

## 9. Pengembangan

Recipe dapat dikembangkan dengan menambahkan inversi 1D atau 2D menggunakan perangkat lunak geolistrik seperti pyGIMLi atau ResIPy, visualisasi penampang resistivitas, serta integrasi dengan data geologi dan hidrogeologi.


```python

import numpy as np  # Mengimpor NumPy untuk operasi numerik
import pandas as pd  # Mengimpor Pandas untuk mengelola tabel data
import matplotlib.pyplot as plt  # Mengimpor Matplotlib untuk membuat grafik

# Membuat data simulasi pengukuran geolistrik
data = {  # Menyimpan data dalam bentuk dictionary
    "Pengukuran": [1, 2, 3, 4, 5],  # Nomor pengukuran
    "Arus_A": [0.20, 0.25, 0.30, 0.22, 0.28],  # Arus dalam ampere
    "Beda_Potensial_V": [0.10, 0.15, 0.18, 0.11, 0.16],  # Tegangan dalam volt
    "Faktor_Geometri_m": [10, 12, 15, 11, 14]  # Faktor geometri dalam meter
}  # Mengakhiri definisi data simulasi

df = pd.DataFrame(data)  # Mengubah data menjadi DataFrame

# Memeriksa kelengkapan data
print("Data pengukuran geolistrik:")  # Menampilkan judul tabel
print(df)  # Menampilkan seluruh data pengukuran

# Memastikan nilai masukan valid
kolom = ["Arus_A", "Beda_Potensial_V", "Faktor_Geometri_m"]  # Menentukan kolom numerik
df[kolom] = df[kolom].apply(pd.to_numeric, errors="coerce")  # Mengubah nilai menjadi numerik
df.replace([np.inf, -np.inf], np.nan, inplace=True)  # Mengubah nilai tak hingga menjadi data kosong

valid = (  # Menentukan pengukuran yang dapat dihitung
    df[kolom].notna().all(axis=1)  # Memastikan semua nilai tersedia
    & (df["Arus_A"] > 0)  # Memastikan arus lebih besar dari nol
    & (df["Faktor_Geometri_m"] > 0)  # Memastikan faktor geometri positif
)

df = df.loc[valid].copy()  # Mempertahankan data dengan nilai masukan yang valid

# Menghitung resistivitas semu
df["Resistivitas_Semu_Ohm_m"] = (  # Membuat kolom hasil perhitungan
    df["Faktor_Geometri_m"]  # Mengambil faktor geometri
    * df["Beda_Potensial_V"]  # Mengalikan dengan beda potensial
    / df["Arus_A"]  # Membagi dengan arus listrik
)  # Menghasilkan resistivitas semu dalam ohm meter

# Menampilkan hasil perhitungan
print("\nHasil perhitungan resistivitas semu:")  # Menampilkan judul hasil
print(df)  # Menampilkan tabel beserta hasil perhitungan

# Menghitung statistik deskriptif
print("\nStatistik resistivitas semu:")  # Menampilkan judul statistik
print(df["Resistivitas_Semu_Ohm_m"].describe())  # Menampilkan statistik ringkasan

# Membuat grafik resistivitas semu
plt.figure(figsize=(8, 5))  # Menentukan ukuran gambar
plt.plot(  # Membuat grafik garis
    df["Pengukuran"],  # Menggunakan nomor pengukuran pada sumbu X
    df["Resistivitas_Semu_Ohm_m"],  # Menggunakan resistivitas semu pada sumbu Y
    marker="o"  # Memberikan penanda pada setiap titik
)  # Mengakhiri perintah grafik
plt.xlabel("Nomor Pengukuran")  # Memberikan label sumbu X
plt.ylabel("Resistivitas Semu (Ω m)")  # Memberikan label sumbu Y
plt.title("Distribusi Resistivitas Semu - Data Simulasi")  # Memberikan judul grafik
plt.grid(True)  # Menampilkan garis bantu
plt.tight_layout()  # Merapikan tata letak grafik
plt.savefig("grafik_resistivitas_se mu.png".replace(" ", ""))  # Menyimpan grafik ke file PNG
plt.show()  # Menampilkan grafik

# Menyimpan hasil perhitungan ke file CSV
df.to_csv("hasil_resistivitas_geolistrik.csv", index=False)  # Menyimpan tabel tanpa indeks
print("\nHasil disimpan ke hasil_resistivitas_geolistrik.csv")  # Menampilkan informasi penyimpanan
```
